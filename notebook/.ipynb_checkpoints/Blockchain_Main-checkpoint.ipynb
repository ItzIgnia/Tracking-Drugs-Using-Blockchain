```python
import web3
web3.__version__
```

```python
from web3 import Web3

w3 = Web3(
    Web3.HTTPProvider("http://127.0.0.1:8545")
)

print("Connected:", w3.is_connected())

print("Chain ID:", w3.eth.chain_id)
print("Block Number:", w3.eth.block_number)
```

```python
accounts = w3.eth.accounts

for i, account in enumerate(accounts):
    print(f"Account {i}: {account}")
    
```

```python
# importing artifact path so as to work with the ABI ( Application Binary Interface which help python to interact with solidity code) Bytecode is the EVM bytecode that runs on Ethereum mainnet 
```

```python
import json

artifact_path = "../artifacts/contracts/DrugTracking.sol/DrugTracking.json"

with open(artifact_path, "r") as file:
    artifact = json.load(file)

abi = artifact["abi"]
bytecode = artifact["bytecode"]

print("ABI loaded:", len(abi) > 0)
print("Bytecode loaded:", len(bytecode) > 0)
```

```python
DrugTracking = w3.eth.contract(
    abi=abi,
    bytecode=bytecode
)

print("Contract object created successfully")
```

```python
deployer = w3.eth.accounts[0]
print("Deploying from:", deployer)
```

```python
tx_hash = DrugTracking.constructor().transact({
    "from": deployer
})

print("Transaction Hash:", tx_hash.hex())

receipt = w3.eth.wait_for_transaction_receipt(tx_hash)

print("Deployment successful!")
print("Contract Address:", receipt.contractAddress)
print("Block Number:", receipt.blockNumber)
```

```python
contract_address = receipt.contractAddress
print(contract_address)

drug_contract = w3.eth.contract(
    address=contract_address,
    abi=abi
)

print("Python connected to DrugTracking contract!")
```

```python
print(
    drug_contract.functions.drugExists("DRUG-001").call()
)
```
